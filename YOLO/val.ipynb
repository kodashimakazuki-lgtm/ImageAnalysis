```python
from ultralytics import YOLO

# 1. 学習済みモデルの読み込み（最も精度が高かった重みファイル）
model = YOLO("best.pt")

# 2. 検証データ（dataset.yamlで指定されたvalデータ）で精度評価を実行
metrics = model.val(data="dataset.yaml")

# 3. 主要な数値指標を表示
print(f"mAP50-95 (総合精度)     : {metrics.box.map:.3f}")    # 厳しい基準での総合精度
print(f"mAP50 (位置ズレ許容精度) : {metrics.box.map50:.3f}")  # 一般的な基準での精度
print(f"Precision (適合率)      : {metrics.box.mp:.3f}")     # 適合率（誤検知の少なさ）
print(f"Recall (再現率)         : {metrics.box.mr:.3f}")     # 再現率（見落としの少なさ）


```
