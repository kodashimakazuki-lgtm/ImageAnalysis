```python
import os
import random
import shutil

# --- 設定 ---
source_dir = r"C:\Users\01A1561\Desktop\imageAnalysis\bottle_output\data\obj"  # VoTTから出力されたデータフォルダ
output_dir = r"C:\Users\01A1561\Desktop\imageAnalysis\bottle_output\data\dataset"              # 保存先フォルダ
split_ratio = 0.8                   # 学習データの割合 (80%)

# フォルダ作成
for folder in ["images/train", "images/val", "labels/train", "labels/val"]:
    os.makedirs(os.path.join(output_dir, folder), exist_ok=True)

# 画像ファイルのリストを取得 (.jpg や .png など)
images = [f for f in os.listdir(source_dir) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
random.shuffle(images)

# 分割位置の計算
split_index = int(len(images) * split_ratio)
train_images = images[:split_index]
val_images = images[split_index:]

def move_files(file_list, subtype):
    for img_name in file_list:
        base_name = os.path.splitext(img_name)[0]
        txt_name = base_name + ".txt"
        
        src_img = os.path.join(source_dir, img_name)
        src_txt = os.path.join(source_dir, txt_name)
        
        # 画像と対応するテキストが存在する場合のみコピー
        if os.path.exists(src_txt):
            shutil.copy(src_img, os.path.join(output_dir, f"images/{subtype}", img_name))
            shutil.copy(src_txt, os.path.join(output_dir, f"labels/{subtype}", txt_name))

# 実行
move_files(train_images, "train")
move_files(val_images, "val")
print("データセットの分割が完了しました！")
```
